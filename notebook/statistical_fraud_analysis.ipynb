```python
# ============================================================
# PHASE 5: HYPOTHESIS TESTING
# ============================================================

from scipy.stats import ttest_ind

print("========== PHASE 5: HYPOTHESIS TESTING ==========")

# Separate legitimate and fraudulent transaction amounts
legitimate_amount = df[df["isFraud"] == 0]["TransactionAmt"]
fraud_amount = df[df["isFraud"] == 1]["TransactionAmt"]

# H0: Transaction amount has no significant difference between
# fraudulent and legitimate transactions.
# H1: Transaction amount significantly differs between them.

t_stat, p_value = ttest_ind(
    legitimate_amount,
    fraud_amount,
    equal_var=False
)

print("\n--- Welch's Independent T-Test ---")
print("T-statistic:", round(t_stat, 4))
print("P-value:", p_value)

alpha = 0.05

print("\nSignificance level (alpha):", alpha)

if p_value < alpha:
    print("Result: Reject H0")
    print("Conclusion: Transaction amount shows a statistically significant")
    print("difference between fraudulent and legitimate transactions.")
else:
    print("Result: Fail to reject H0")
    print("Conclusion: No statistically significant difference was found.")

print("\n--- Business Interpretation ---")
print("Fraudulent transactions have a slightly higher average amount,")
print("but transaction amount alone should not be treated as a fraud decision rule.")

print("\n========== PHASE 5 COMPLETED ==========")
```

```python
# ============================================================
# PHASE 4: PATTERN ANALYSIS
# ============================================================

print("========== PHASE 4: PATTERN ANALYSIS ==========")

# 1. Transaction Amount vs Fraud
print("\n--- Transaction Amount vs Fraud ---")
print(
    df.groupby("isFraud")["TransactionAmt"]
    .agg(["mean", "median"])
)

# 2. Time of Day vs Fraud
df["Hour"] = (df["TransactionDT"] % 86400) // 3600

hourly_fraud = df.groupby("Hour")["isFraud"].mean() * 100

print("\n--- Fraud Rate by Hour ---")
print(hourly_fraud.round(2))

# 3. Product Category vs Fraud
product_fraud = df.groupby("ProductCD")["isFraud"].mean() * 100

print("\n--- Fraud Rate by Product Category ---")
print(product_fraud.round(2))

# 4. Transaction Frequency
frequency = df.groupby("card1")["TransactionID"].transform("count")

df["TransactionFrequency"] = frequency

frequency_by_fraud = df.groupby("isFraud")["TransactionFrequency"].mean()

print("\n--- Average Transaction Frequency by Fraud Status ---")
print(frequency_by_fraud.round(2))

print("\n========== PHASE 4 COMPLETED ==========")
```

```python
# ============================================================
# Z-SCORE OUTLIER DETECTION
# ============================================================

print("\n--- Z-SCORE METHOD ---")

mean_amount = df["TransactionAmt"].mean()
std_amount = df["TransactionAmt"].std()

z_scores = (df["TransactionAmt"] - mean_amount) / std_amount

zscore_outliers = df[z_scores.abs() > 3]

print("Z-score threshold: |Z| > 3")
print("Number of Z-score outliers:", len(zscore_outliers))
print("Z-score outlier percentage:",
      round(len(zscore_outliers) / len(df) * 100, 2), "%")

print("\n--- Highest Z-Score Transactions ---")

result = df[["TransactionID", "TransactionAmt", "isFraud"]].copy()
result["ZScore"] = z_scores

display(
    result.sort_values("ZScore", ascending=False).head(10)
)
```

```python
# ============================================================
# PHASE 3: OUTLIER DETECTION
# ============================================================

print("========== PHASE 3: OUTLIER DETECTION ==========")

Q1 = df["TransactionAmt"].quantile(0.25)
Q3 = df["TransactionAmt"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("\n--- IQR Method ---")
print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))
print("Lower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))

outliers = df[
    (df["TransactionAmt"] < lower_bound) |
    (df["TransactionAmt"] > upper_bound)
]

print("Number of amount outliers:", len(outliers))
print("Outlier percentage:",
      round(len(outliers) / len(df) * 100, 2), "%")

print("\n--- Highest Transaction Amounts ---")
display(
    df[["TransactionID", "TransactionAmt", "isFraud"]]
    .sort_values("TransactionAmt", ascending=False)
    .head(10)
)
```
